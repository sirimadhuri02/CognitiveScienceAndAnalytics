```python
import nltk
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.corpus import wordnet
from nltk import pos_tag, word_tokenize

# Download resources
nltk.download("punkt")
nltk.download("averaged_perceptron_tagger_eng")
nltk.download("wordnet")
nltk.download("omw-1.4")
nltk.download('punkt_tab')

# Initialize tools
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

# Convert POS tags to WordNet-compatible tags
def get_wordnet_pos(tag):
    if tag.startswith('J'):
        return wordnet.ADJ
    elif tag.startswith('V'):
        return wordnet.VERB
    elif tag.startswith('N'):
        return wordnet.NOUN
    elif tag.startswith('R'):
        return wordnet.ADV
    else:
        return wordnet.NOUN

# Classify ambiguity type
def classify_ambiguity(stems, lemmas, pos_tags):
    results = []

    # Morphological ambiguity: different words → same stem
    for stem, words in stems.items():
        if len(words) > 1:
            results.append((stem, words, "Morphological Ambiguity"))

    # POS ambiguity: same lemma used with different POS
    lemma_pos_map = {}
    for word, tag in pos_tags:
        wn_pos = get_wordnet_pos(tag)
        lemma = lemmatizer.lemmatize(word, wn_pos)
        lemma_pos_map.setdefault(lemma, set()).add(wn_pos)

    for lemma, pos_set in lemma_pos_map.items():
        if len(pos_set) > 1:
            results.append((lemma, pos_set, "POS Ambiguity"))

    return results

# Sentences
sentences = [
    "The manager managing the team efficiency was praised for his leadership",
    "After attending the meetings, she recorded the minutes and noted important decisions",
    "The cats chasing the mice were chased by other animals in the wild",
    "The lead researcher leads the team toward finding new solutions to lead contamination",
    "Her readings about reading comprehension helped her understand the material better",
   "Flying planes can be dangerous if the weather changes suddenly.",
"The old man the boats near the shore during the storm.",
"She saw her duck before it jumped into the water.",
"the band saw the saw and decided to fix it immediately.",
 "I will book the book before the event begins. "
]

# Run
for sent in sentences:
    print(f"\nSentence: {sent}")
    tokens = word_tokenize(sent)
    pos_tags = pos_tag(tokens, lang="eng")

    stems, lemmas = {}, {}

    print("\nWord | Stem | Lemma | POS")
    print("-"*45)

    for word, tag in pos_tags:
        stem = stemmer.stem(word)
        lemma = lemmatizer.lemmatize(word, get_wordnet_pos(tag))
        print(f"{word:10} | {stem:10} | {lemma:10} | {tag}")

        stems.setdefault(stem, set()).add(word)
        lemmas.setdefault(lemma, set()).add(word)

    # Classify ambiguities
    ambiguities = classify_ambiguity(stems, lemmas, pos_tags)

    if ambiguities:
        print("\n🔎 Ambiguities Found:")
        for item in ambiguities:
            print(" -", item)
    else:
        print("\n✅ No ambiguity detected")
```
