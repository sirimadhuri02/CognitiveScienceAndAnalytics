```python
import pandas as pd
import nltk
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.corpus import wordnet
```

```python
# Download resources (safe to run multiple times, only downloads if missing)
nltk.download('punkt')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
# Handle POS tagger resource depending on NLTK version
try:
    nltk.download('averaged_perceptron_tagger_eng')
except:
    nltk.download('averaged_perceptron_tagger')
```

```python
# Function to map POS tags to WordNet POS
def get_wordnet_pos(treebank_tag):
    if treebank_tag.startswith('J'):
        return wordnet.ADJ
    elif treebank_tag.startswith('V'):
        return wordnet.VERB
    elif treebank_tag.startswith('N'):
        return wordnet.NOUN
    elif treebank_tag.startswith('R'):
        return wordnet.ADV
    else:
        return wordnet.NOUN  # Default to noun
```

```python
pos_explanations = {
    "NN": "Noun, singular",
    "NNS": "Noun, plural",
    "VB": "Verb, base form",
    "VBD": "Verb, past tense",
    "VBG": "Verb, gerund/present participle",
    "VBN": "Verb, past participle",
    "VBP": "Verb, non-3rd person singular present",
    "VBZ": "Verb, 3rd person singular present",
    "JJ": "Adjective",
    "JJR": "Adjective, comparative",
    "JJS": "Adjective, superlative",
    "RB": "Adverb",
    "RBR": "Adverb, comparative",
    "RBS": "Adverb, superlative",
}
```

```python

# Input words
words = ["studies", "running", "better", "cared", "flies"]
```

```python
# Initialize stemmer and lemmatizer
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()
```

```python
# POS tagging
pos_tags = nltk.pos_tag(words)
```

```python
# Prepare results
results = []
for word, pos in pos_tags:
    stemmed = stemmer.stem(word)
    lemma_simple = lemmatizer.lemmatize(word)  # Lemma without POS
    lemma_pos = lemmatizer.lemmatize(word, get_wordnet_pos(pos))  # Lemma with POS
    results.append([word, stemmed, lemma_simple, lemma_pos])
```

```python
# Create DataFrame
df = pd.DataFrame(results, columns=["Word", "Stemming", "Lemmatization (No POS)", "Lemmatization (With POS)"])

# Display table
display(df)
```

```python
sentence = "The children were running and studies were completed with better solutions while caring for the flies."
words2 = nltk.word_tokenize(sentence)
pos_tags2 = nltk.pos_tag(words2)
```

```python

results2 = []
for word, pos in pos_tags2:
    if word.isalpha():  # skip punctuation
        stemmed = stemmer.stem(word)
        lemma_simple = lemmatizer.lemmatize(word)
        lemma_pos = lemmatizer.lemmatize(word, get_wordnet_pos(pos))
        results2.append([word, pos, stemmed, lemma_simple, lemma_pos])
```

```python
df2 = pd.DataFrame(
    results2,
    columns=["Word", "POS", "Stemming", "Lemma (No POS)", "Lemma (With POS)"]
)

print("\n=== Example 2: Sentence with POS Explanations ===")
display(df2)
```
