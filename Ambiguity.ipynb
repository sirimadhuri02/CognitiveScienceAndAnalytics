```python
import pandas as pd
import nltk
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.corpus import wordnet

# Downloads
nltk.download('punkt')
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('punkt_tab')
try:
    nltk.download('averaged_perceptron_tagger_eng')
except:
    nltk.download('averaged_perceptron_tagger')

# POS explanation dictionary
pos_explanations = {
    "NN": "Noun, singular",
    "NNS": "Noun, plural",
    "NNP": "Proper noun, singular",
    "NNPS": "Proper noun, plural",
    "VB": "Verb, base form",
    "VBD": "Verb, past tense",
    "VBG": "Verb, gerund/present participle",
    "VBN": "Verb, past participle",
    "VBP": "Verb, non-3rd person singular present",
    "VBZ": "Verb, 3rd person singular present",
    "JJ": "Adjective",
    "JJR": "Adjective, comparative",
    "JJS": "Adjective, superlative",
    "RB": "Adverb",
    "RBR": "Adverb, comparative",
    "RBS": "Adverb, superlative",
    "PRP": "Pronoun",
    "PRP$": "Possessive pronoun",
    "DT": "Determiner",
    "IN": "Preposition",
    "CC": "Coordinating conjunction",
    "UH": "Interjection"
}

# POS mapping for WordNet lemmatizer
def get_wordnet_pos(treebank_tag):
    if treebank_tag.startswith('J'):
        return wordnet.ADJ
    elif treebank_tag.startswith('V'):
        return wordnet.VERB
    elif treebank_tag.startswith('N'):
        return wordnet.NOUN
    elif treebank_tag.startswith('R'):
        return wordnet.ADV
    else:
        return wordnet.NOUN

# Initialize tools
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

# Sentences
sentence1 = "The manager managing the team effeciently was praised for his leadership."
sentence2 = "After attending the meetings, she recorded the minutes and noted important decisions"
sentence3 = "The cats chasing the mice were chased by other animals in the wild"
sentence4 = "the lead researcher leads the team toward finding new solutions to lead contamination"
sentence5 = "Her readings about the reading comprehention helped her understand the material better."

sentences = [sentence1, sentence2, sentence3, sentence4, sentence5]

# Process each sentence and assign to separate DataFrames
def process_sentence(sentence):
    words = nltk.word_tokenize(sentence)
    pos_tags = nltk.pos_tag(words)
    results = []
    for word, pos in pos_tags:
        if word.isalpha():
            stemmed = stemmer.stem(word)
            lemma_simple = lemmatizer.lemmatize(word)
            lemma_pos = lemmatizer.lemmatize(word, get_wordnet_pos(pos))
            pos_expl = pos_explanations.get(pos, "Other/Unknown")
            results.append([word, pos, pos_expl, stemmed, lemma_simple, lemma_pos])
    df = pd.DataFrame(
        results,
        columns=["Word", "POS", "POS Explanation", "Stemming", "Lemma (No POS)", "Lemma (With POS)"]
    )
    return df

df1 = process_sentence(sentence1)
df2 = process_sentence(sentence2)
df3 = process_sentence(sentence3)
df4 = process_sentence(sentence4)
df5 = process_sentence(sentence5)

# Display each DataFrame
print("\n=== DataFrame for Sentence 1 ===")
display(df1)

print("\n=== DataFrame for Sentence 2 ===")
display(df2)

print("\n=== DataFrame for Sentence 3 ===")
display(df3)

print("\n=== DataFrame for Sentence 4 ===")
display(df4)

print("\n=== DataFrame for Sentence 5 ===")
display(df5)
```
