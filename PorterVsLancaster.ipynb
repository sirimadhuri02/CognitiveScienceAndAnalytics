```python
import nltk
from nltk.stem import PorterStemmer, LancasterStemmer
from nltk.tokenize import word_tokenize
import pandas as pd
```

```python
nltk.download("punkt")
nltk.download("punkt_tab")
```

```python
text1 = "He was running and playing with his friends."
text2 = "The universities are organizing several programs."
```

```python
text1_tok = word_tokenize(text1)
text2_tok = word_tokenize(text2)
```

```python
ps = PorterStemmer()
ls = LancasterStemmer()
```

```python
stemmed_text1_porter = [ps.stem(word) for word in text1_tok]
stemmed_text2_porter = [ps.stem(word) for word in text2_tok]
```

```python
stemmed_text1_lancaster = [ls.stem(word) for word in text1_tok]
stemmed_text2_lancaster = [ls.stem(word) for word in text2_tok]
```

```python
print("Original Text1:", text1_tok)
print("Porter Stemmed Text1:", stemmed_text1_porter)
print("Lancaster Stemmed Text1:", stemmed_text1_lancaster)
```

```python
print("\nOriginal Text2:", text2_tok)
print("Porter Stemmed Text2:", stemmed_text2_porter)
print("Lancaster Stemmed Text2:", stemmed_text2_lancaster)
```

```python
words = ["running", "playing", "universities", "programs", "dogs", "happiness"]
```

```python
data = {
    "Original": words,
    "Porter Stemmer": [ps.stem(w) for w in words],
    "Lancaster Stemmer": [ls.stem(w) for w in words]
}
```

```python
df = pd.DataFrame(data)
print("\nComparison Table:\n", df)
```
